## 도서 목록 데이터 추출 및 구조화

```python
from bs4 import BeautifulSoup

html_books = """
<div class='library'>
<div class='book' id='b-101'>
<h3 class='book-title'>딥러닝 텐서플로 마스터</h3>
<p class='book-author'>김작가</p>
</div>
<div class='book' id='b-102'>
<h3 class='book-title'>이것이 진짜 R 프로그래밍</h3>
<p class='book-author'>박통계</p>
</div>
<div class='book' id='b-103'>
<h3 class='book-title'>자연어 처리 입문</h3>
<p class='book-author'>최이과</p>
</div>
</div>
"""

soup = BeautifulSoup(html_books, 'html.parser')

books_info = []

for book in soup.select(".book"):
    title = book.select_one(".book-title").string
    author = book.select_one(".book-author").string

    books_info.append({"title": title, "author" : author})


print("=== 추출된 도서 목록 ===")
print(books_info)
```

## 뉴스 링크 추출

```python
from bs4 import BeautifulSoup

html_news = """
<ul class='news-section'>
<li class='news-item'><a href='https://news.example.com/ai-1'>인공지능 규제법 통과, IT 업계 파장</a></li>
<li class='news-item'><a href='https://news.example.com/dev-2'>파이썬 3.12 릴리즈, 무엇이 달라졌나?</a></li>
<li class='news-item'><a href='https://news.example.com/sec-3'>오픈소스 라이브러리 보안 취약점 경고</a></li>
</ul>
"""

soup = BeautifulSoup(html_news, 'html.parser')

news_list = []

for news in soup.select(".news-item"):
    headline = news.select_one('a').string
    url = news.select_one('a').get('href')

    news_list.append({"headline":headline, "url":url})

print("=== 수집된 뉴스 헤드라인 ===")
print(news_list)
```

## SQL 데이터 조작

```python
import sqlite3

create_users = '''
    create table users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        email TEXT,
        level TEXT
    )
'''

insert_user = '''
    insert into users (name, email, level) VALUES (?, ?, ?)
'''

select_gold_users = '''
    select name, email FROM users WHERE level = "gold"
'''

user_list = [
    ('홍길동', 'hong@example.com', 'gold'),
    ('이영희', 'lee@example.com', 'silver'),
    ('김철수', 'kim@example.com', 'gold')
]

conn = sqlite3.connect('info.db')

cursor = conn.cursor()
cursor.execute(create_users)

cursor.executemany(insert_user, user_list)
conn.commit()

cursor.execute(select_gold_users)
gold_users = cursor.fetchall()

print('=== Gold 등급 회원 목록 ===')
print(gold_users)

# 6. 연결 종료
conn.close()
```

# SQL 데이터 복원

```python
import pandas as pd
import sqlite3

# 실습 데이터프레임 생성
df_members = pd.DataFrame({
'member_id': ['M01', 'M02', 'M03'],
'name': ['이민수', '김영희', '박준형'],
'score': [85, 95, 78]
})

conn = sqlite3.connect('users.db')
df_members.to_sql('members', conn, if_exists='replace', index=False)
print("DataFrame을 SQLite 데이터베이스에 저장 완료")

conn.close()
```

```python
conn = sqlite3.connect('users.db')

query = "SELECT * FROM members"
df_read = pd.read_sql_query(query, conn)
print("======== DB 복원 내용========")
print(df_read)

conn.close()
```
