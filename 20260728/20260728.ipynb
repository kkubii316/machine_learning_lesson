```python
import requests

url = 'https://jsonplaceholder.typicode.com/posts'

response = requests.get(url)
print(type(response))

print(response.status_code)

if response.status_code == 200 :        # 정상
    print('정상응답')
    print(response.text[:200])          # text(JSON)
else:       # 에러(4xx, 5xx)
    print('정상적인 수신이 되지 않았습니다.')
    print(f'오류발생: {response.status_code}')
```

```python
data_obj = response.json()

print(type(data_obj))

for post in data_obj:
    print(f'title : {post['title']}')
    print(f'body ; {post['body'][:10]}')
    print('-'*20)
```

## 웹 스크래핑

```python
# find(tag) : tag의 element 하나를 추출
# find_all(tag) : tag에 해당하는 elements 모두 추출
# select(css 선택자) : find_all과 같은 역
# select_one(css 선택자) : find와 같은 역할
# css 선택자는 태그의 class 속성값으로 선택
```

```python
!pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

html_doc = """
<html>
<head><title>Example Page</title></head>
<body>
<h1>Welcome to Web Scraping</h1>
<p class="description">This is an example paragraph.</p>
<a href="https://example.com">Visit Example</a>
<a href="https://example1.com">Visit Example1</a>
<a href="https://example2.com">Visit Example2</a>
</body>
</html>
"""

soup =  BeautifulSoup(html_doc, 'html.parser')

print(soup.title)        # 객체의 속성값을 이용해서 데이터 추출
print(soup.title.string)
print(soup.h1.string)
print(soup.a) 
print(soup.a.string)   
print(soup.a['href'])     
```

```python
a_tag = soup.find('a')
print(a_tag)

a_tags = soup.find_all('a')
print(a_tags)
```

```python
links = []
for tag in a_tags:
    links.append(tag['href'])

print(links)
```

```python
desc_tag  = soup.select('.description')
print(desc_tag)
desc_tag = soup.select('a')
print(desc_tag)
```

```python
"""
css 선택자
*       : 전체 선택자
태그    : 태그 선택자 a, p, div
.       : class 선택자 .desc
#       : id 선택자 #one -> <div id="one>...</div>
"""
```

```python
url_2 = 'https://news.ycombinator.com/'

res = requests.get(url_2)

print(res.status_code)
print(res.text[:500])

soup2 = BeautifulSoup(res.text, 'html.parser')

print(soup2.find_all('a'))

for item in soup2.find_all('a'):
    href_text = item.get('href')
    title_text = item.get_text()
    if href_text and title_text:
        print(title_text, href_text)
```

## SQLite

```python
# SQLite : 관계형데이터베이스
# cursor : SQL을 실행 시키기 위한 도구(객체)
# 데이터 입력은 행 단위로 입력됨
# commit() 메서드 : 정상처리 확인 및 저장
# rollback() 메서드 : 문제 발생 확인 및 되돌림
```

```python
import sqlite3

conn = sqlite3.connect('first.db')
print('데이터베이스 연결 성공!')

conn.close()
```

```python
sql_create = '''
create table user (
    id integer primary key autoincrement,
    name text not null,
    age integer,
    city text
)
'''

conn = sqlite3.connect('first.db')
print('conn ok!!')

cursor = conn.cursor()

cursor.execute(sql_create)
print('create Table ok!!!')

conn.close()
```

```python
sql_insert = '''
insert into user (name, age, city) values(?, ?, ?)
'''

conn = sqlite3.connect('first.db')

cursor = conn.cursor()

data = [
    ('hot', 25, 'Seoul'),
    ('king', 21, 'Suwon'),
    ('son', 30, 'London'),
]

cursor.executemany(sql_insert, data)

conn.commit()

conn.close()
```

```python
sql_select = 'select * from user'
sql_select1 = 'select name, city, age from user'
sql_select2 = 'select * from user where id>2 and age>25'

conn = sqlite3.connect('first.db')
cursor = conn.cursor()

cursor.execute(sql_select2)
rows = cursor.fetchall()

print(rows)

conn.close()
```

```python
sql_update = 'update user set age=?, name=? where id=?'

conn = sqlite3.connect('first.db')
cursor = conn.cursor()

cursor.execute(sql_update, (30, 'KING', 1))

conn.commit()
conn.close()
```

```python
sql_delete = 'delete from user where id=?'

conn = sqlite3.connect('first.db')
cursor = conn.cursor()

cursor.execute(sql_delete, (1,))

conn.commit()
conn.close()
```
